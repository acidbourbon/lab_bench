```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
data = np.loadtxt("mheil_SiPM_shaper_tot.csv",delimiter=",")

x = data[:,0]
y = data[:,1]
```

```python
plt.scatter(x,y)
plt.plot(x,y)
plt.title("SiPM amplitude vs shaper output tot")
plt.xlabel("voltage (mV)")
plt.ylabel("tot (ns)")
plt.show()
```
