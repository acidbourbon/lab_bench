```python
##################################################
##               arbitrary pulse                ##
##################################################


period = 1e-6


x = np.arange(0,period,1e-9)

Q = 150e-12
tau1 = 4e-9
tau2 = 10e-9
delay=100e-9
y = detector_signal_function(x,Q,tau1,tau2,delay)

plt.plot(x*1e9,y*1e3)
plt.xlabel("ns")
plt.ylabel("mV")
plt.show()

awg.program_trace(x,y,trace=1,period=period)
```
